Provide paths to your already downloaded BEAT2 English data and external text embedding file.

```python
import sys, subprocess
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
assert (ROOT / "modules").is_dir(), f"Open this notebook from TalkFlow repository root or notebooks/: {ROOT}"
```

```python
DATA_ROOT = Path('/path/to/BEAT2/beat_english_v2.0.0')
OUTPUT = ROOT / 'data' / 'beat2_processed'
FASTTEXT_VEC = Path('/path/to/fasttext.vec')
DEVICE = 'cuda'
SCRIPT = 'preprocess_beat2.py'
ARGS = ['all', '--data-root', str(DATA_ROOT), '--out-dir', str(OUTPUT), '--fasttext-vec', str(FASTTEXT_VEC), '--device', DEVICE]
```

```python
command = [sys.executable, str(ROOT / "scripts" / SCRIPT), *ARGS]
print("Running:", " ".join(map(str, command)))
subprocess.run(command, check=True, cwd=ROOT)
```
