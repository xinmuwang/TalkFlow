# Stage I — Autoencoder Evaluation

Load local Stage-I weights to evaluate reconstruction and optionally export examples.

```python
import sys, subprocess
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
assert (ROOT / "modules").is_dir(), f"Open this notebook from TalkFlow repository root or notebooks/: {ROOT}"
```

```python
DATA = ROOT / 'data' / 'beat2_processed'
AE_CHECKPOINT = Path('/path/to/ae_checkpoint.pt')
SAVE_DIR = ROOT / 'outputs' / 'ae_eval'
SCRIPT = 'eval_ae.py'
ARGS = ['--data', str(DATA), '--checkpoint', str(AE_CHECKPOINT),
        '--out-dir', str(SAVE_DIR), '--save-examples', '5']
```

```python
command = [sys.executable, str(ROOT / "scripts" / SCRIPT), *ARGS]
print("Running:", " ".join(map(str, command)))
subprocess.run(command, check=True, cwd=ROOT)
```
