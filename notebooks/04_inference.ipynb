# Three-Stream Inference

Generate SMPL-X NPZ sequences using local checkpoints. Optional postprocessing is enabled by default.

```python
import sys, subprocess
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
assert (ROOT / "modules").is_dir(), f"Open this notebook from TalkFlow repository root or notebooks/: {ROOT}"
```

```python
DATA = ROOT / 'data' / 'beat2_processed'
AE_CHECKPOINT = Path('/path/to/ae_checkpoint.pt')
FLOW_CHECKPOINT = Path('/path/to/flow_checkpoint.pt')
SAVE_DIR = ROOT / 'outputs' / 'generated_npz'
ENABLE_SMOOTH = True
SCRIPT = 'infer.py'
ARGS = ['--data', str(DATA), '--ae-checkpoint', str(AE_CHECKPOINT),
        '--flow-checkpoint', str(FLOW_CHECKPOINT), '--out-dir', str(SAVE_DIR)]
if not ENABLE_SMOOTH:
    ARGS += ['--no-smooth']
```

```python
command = [sys.executable, str(ROOT / "scripts" / SCRIPT), *ARGS]
print("Running:", " ".join(map(str, command)))
subprocess.run(command, check=True, cwd=ROOT)
```
