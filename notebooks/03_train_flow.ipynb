# Stage II — Flow Training

Choose a training stage and provide locally stored AE/Flow checkpoints.

```python
import sys, subprocess
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
assert (ROOT / "modules").is_dir(), f"Open this notebook from TalkFlow repository root or notebooks/: {ROOT}"
```

```python
DATA = ROOT / 'data' / 'beat2_processed'
AE_CHECKPOINT = Path('/path/to/ae_checkpoint.pt')
PREVIOUS_FLOW_CHECKPOINT = None  # Path('/path/to/previous_flow_checkpoint.pt')
STAGE = 1
SAVE_DIR = ROOT / 'checkpoints' / 'flow'
SCRIPT = 'train_flow.py'
ARGS = ['--data', str(DATA), '--ae-checkpoint', str(AE_CHECKPOINT),
        '--stage', str(STAGE), '--out-dir', str(SAVE_DIR)]
if PREVIOUS_FLOW_CHECKPOINT is not None:
    ARGS += ['--resume', str(PREVIOUS_FLOW_CHECKPOINT)]
```

```python
command = [sys.executable, str(ROOT / "scripts" / SCRIPT), *ARGS]
print("Running:", " ".join(map(str, command)))
subprocess.run(command, check=True, cwd=ROOT)
```
